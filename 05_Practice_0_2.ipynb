```python
# Problem Statement
# Take a string as input from the user.
# Display
# Number of vowels
# Number of consonants
# Number of digits
# Number of spaces
# # Ip1: Python3 is Fun
# # op1: vowels - 3
# # consonants - 8
# # digits - 1
# # spaces - 2
```

```python
s = input()
sl = s.lower()
vowels = 0
consonants = 0
digits = 0
spaces = 0
for c in sl:
 if c in "aeiou":
  vowels += 1
 elif c >= "a" and c <= "z":
  consonants += 1
 elif c >="0" and c <= "9":
  digits += 1
 elif c == " ":
  spaces += 1
print("Vowels -", vowels)
print("Consonants -", consonants)
print("Digits -", digits)
print("Spaces -", spaces)
```

```python
#version2
text = input()
vowels = 0
consonants = 0
digits = 0
spaces = 0
for ch in text:
 if ch.lower() in "aeiou":
  vowels += 1
 elif ch.isalpha():
  consonants += 1
 elif ch.isdigit():
  digits += 1
 elif ch == " ":
  spaces += 1
print("Vowels :", vowels)
print("Consonants :", consonants)
print("Digits :", digits)
print("Spaces :", spaces)
```
