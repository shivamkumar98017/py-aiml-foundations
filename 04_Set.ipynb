```python
#Write a code to only store the unique occurances of the elements in the list.
students = [
"Rahul",
"Amit",
"Rahul",
"Neha",
"Amit",
"Rahul"
]
s1 = []
for student in students:
 if student not in s1:
  s1.append(student)
print(s1)
```
