Higher Order function(HOF)

```python
# HOF are functions that:
#  1- accept another function as an argument
#  2 - return another function
```

```python
def add(x,y):
  return x+y

def mul (x,y):
 return x*y

def operation_fun(f,x,y):
  return f(x,y)

ans = operation_fun(add,3,5)
ans
```

```python
ans = operation_fun(mul,3,5)
ans
```

```python
# given a list of integers,find the largest element without using python's buit-in max()function.
numbers = [12,45,7,89,34]

largest = numbers[0]
for num in numbers:
  if num > largest:
    largest = num

print("Largest element:",largest)
```

```python
# Given a list of distinct integer,find the second largest element without shorting the list.
numbers = [10,54,20,78,34]
largest = float('-inf')
second_largest = float('-inf')

for num in numbers:
  if num > largest:
    second_largest = largest
    largest = num

  elif num > second_largest:
    second_largest = num
```

```python
print(largest)
print(second_largest)
```

```python
# Given a list of integers, create a new list conataining only unique elements.
# Maintain the original order of elements.
# input = [10,20,10,30,20,40]
# output = [10,20,30,40]

numbers = [10,20,10,30,20,40]
unique = []

for num in numbers:
  if num not in unique:
    unique.append(num)

print("Unique element:",unique)
```

```python
# Move All zeros to the end

# Given a list of integers, move all the zeros to the end of list while maintaining the relative order of the non-zero elements.

# input:[0,5,0,3,8,0,2]
# output:[5,3,8,2,0,0,0]

numbers = [0,5,0,3,8,0,2]

result = []

zero_count = 0

for num in numbers:
  if num == 0:
    zero_count += 1
  else:
    result.append(num)

for i in range(zero_count):
  result.append(0)

print("Result", result)
```

```python
# Given a list of number, calculate the average and print all elements that are greater than the average.
# (without numpy)

# Input = [10,20,30,40,50]
# output = [40,50]

numbers = [10,20,30,40,50]

total = 0

for num in numbers:
  total+=num

average = total/len(numbers)

result = []

for num in numbers:
  if num > average:
    result.append(num)

print("Average:",average)
print("Elements greater than average:",result)
```

```python
# Given a list containing positive, negative and zero values, create two list:

# positive numbers
# Negative numbers

# ignore zero.

# Input:[10,-5,0,8,-3,12,-7]
# Output:positive[10,8,12]
        #  negative[-5,-3,-7]
```

```python
numbers = [10,-5,0,8,-3,12,-7]
positive = []
negative = []

for num in numbers:
  if num > 0:
    positive.append(num)

  elif num < 0:
    negative.append(num)


print("Positive:",positive)
print("Negative:",negative)
```

```python
# Given two lists, find the elements that appears in both lists.

# list 1: [10,20,30,40]
# list 2: [30,40,50,60]

list1 = [10,20,30,40]
list2 = [30,40,50,60]

common = []

for num in list1:
  if num in list2:
    common.append(num)

print("Common Element:", common)
```
