```python
# Write a function to Print Grade of Every Student
# Rules
# 90+
# A
# 75+
# B
# 60+
# C
# Else
# D
```

```python
def calc_grade(marks):
 if marks > 90:
  grade = "A"
 elif marks > 75:
  grade = "B"
 elif marks > 60:
  grade = "C"
 else:
  grade = "D"

 return grade
```

```python
for student in students:
 print(student["marks"], calc_grade(student["marks"]))
```

```python
# Write a function to print the topper of the class.
def find_topper(students):
  mx_marks = 0
  topper = {}
  for student in students:
    if student["marks"] > mx_marks:
      mx_marks = student["marks"]
      topper = student
  return topper
```

```python
ans = find_topper(students)
print(ans)
```

```python
class Student:
  def __init__(self, roll, name, age, course, marks):
    self.roll = roll
    self.name = name
    self.age = age
    self.course = course
    self.marks = marks

  def show_student(self):
    print(self.roll, self.name, self.age, self.course, self.marks)

s1 = Student(106, "MJ", 30, "Dance", 100)
s1.show_student()
```
