```python
numbers1 = [1,2,3]
numbers2 = [10,20]

pair = []

for n1 in numbers1:
  for n2 in numbers2:
    pair.append((n1,n2))
pair
```

```python
# Do with  using list comprehension
pair = [(n1,n2) for n1 in numbers1 for n2 in numbers2]
pair
```

```python
# Find pairs of numbers from 1 to 5 where both numbers are different and their product is even.
pairs = []
for n1 in range(1,6):
 for n2 in range(1,6):
  if (n1 != n2)&(n1*n2%2==0):
    pairs.append((n1,n2))

pairs
```

```python
# replace negative values with zero in a matrix
data = [
    [10,-5,20],
    [-3,15,-8],
    [25,0,12]
]

replace = [e if e >=0 else 0 for r in data for e in r]
replace
```

```python
# Each inner list represents the price of items in one customer`s cart.
# calculate the total for every customer.

carts = [
    [100,250,50],
    [500,200],
    [120,80,300,100]
]

# Safely calculate the total using built-in sum.
# Note: If you get a TypeError, please run 'del sum' in a new cell first to clear the overridden variable.
total = [sum(c) for c in carts]
total
```
