```python
# Problem statement: an array contains distinct numbers from 1 to n,
# but one number is missing. find the missing number.

# Input: arr = [1,2,4,5,6],n = 6
# output: 3
```

```python
def find_missing(arr,n):
    expected_sum = n*(n+1)//2
    actual_sum = sum(arr)

    return expected_sum-actual_sum

arr = [1,2,4,5,6]
print(find_missing(arr,6))
```

```python
# Find two numbers whose sum equals a tarrget

# problem statements: Given an aaray and a target value,
# return the indices of two distinct elements whose sum equals to target.
# Assume exactly one valid pair exits.

# Input: arr =[2,7,11,15],target = 9
# Output:[0,1]
```

```python
def two_sum(arr, target):
    seen = {}

    for i, num in enumerate(arr):
        required = target - num

        if required in seen:
            return [seen[required], i]

        # store index instead of 1
        seen[num] = i

    return []

arr = [2, 7, 11, 15]
print(two_sum(arr, 9))  # Output: [0, 1]
```

```python
def is_palindrdrome(text):
  cleaned = ""

  for char in text.lower():
    if char.isalnum():
      cleaned+=char

  left = 0
  right = len(cleaned) - 1

  while left < right:
    if cleaned[left] != cleaned[right]:
      return False

    left+=1
    right-=1

  return True

print(is_palindrdrome("A man a plane,a canal:panama"))
print(is_palindrdrome("Hello"))
```

```python
# problem statemets:Two strings are anagrams
# if they contain the same characters with the the same frequencies,regardless of orders.
# Ignore space.

# Input:
# s1:"Listen"
# s2:"Silent"

# output:"Yes"
```

```python
def are_anagrams(s1, s2):
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()

    if len(s1) != len(s2):
        return False

    frequency = {}

    for char in s1:
        frequency[char] = frequency.get(char, 0) + 1

    for char in s2:
        if char not in frequency:
            return False

        frequency[char] -= 1

        if frequency[char] < 0:
            return False

    return True


print(are_anagrams("Listen", "Silent"))
print(are_anagrams("Hello", "World"))
```

```python
# problem statement: Given s string,find the first character that appears only exactly ones.
# Return -1 if they character repeats.

# Input: "aabbcddee"
# output:'c'
```

```python
def first_unique_character(text):
    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    for char in text:
        if frequency[char] == 1:
            return char

    return -1


print(first_unique_character("aabbcddee"))
print(first_unique_character("aabb"))
```

```python
# Problem statement:
# Given two aarays sorted in ascending order,
# merge them into one sorted array without sorting the combine result afterward.

# Input:[1,4,7,9] and [2,3,8,10]
# output:[1,2,3,4,7,8,9,10]
```

```python
def merge_sorted_arrays(a, b):
    i = 0
    j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    while i < len(a):
        result.append(a[i])
        i += 1

    while j < len(b):
        result.append(b[j])
        j += 1

    return result


a = [1, 4, 7, 9]
b = [2, 3, 8, 10]

print(merge_sorted_arrays(a, b))
```

```python
arr = [7, 1, 5, 3, 6, 4]

def largest_diff(arr):
    max_diff = arr[1] - arr[0]

    for i in range(len(arr)):
        for j in range(i + 1, len(arr)):
            if arr[j] - arr[i] > max_diff:
                max_diff = arr[j] - arr[i]

    return max_diff

largest_diff(arr)
```

```python
# Find the Smallest Difference Between Two Sorted Arrays

# Problem statement: Given two arrays sorted in ascending order,
#  find the smallest absolute difference between an element of the first array and an element of the second array.

# Example:
# A = [1, 4, 8, 10]
# B = [2, 5, 9, 12]

# Output:1
```

```python

A = [1, 4, 8, 10]
B = [2, 5, 9, 12]

def find_smallest_diff(arr1, arr2):
    smallest_diff = float('inf')

    for i in range(len(arr1)):
        for j in range(len(arr2)):
            diff = abs(arr1[i] - arr2[j])

            if diff < smallest_diff:
                smallest_diff = diff

    return smallest_diff

print(find_smallest_diff(A, B))
```

```python
# Problem:Find the greatest common divisor(GCD) of two positive integers
# using the Euclidean alogorithm.

# input:48,18
# output:6
```

```python
arr = [18, 48]

def find_gcd(arr):
    greatest_divisor = 1

    for i in range(1, min(arr) + 1):
        if arr[0] % i == 0 and arr[1] % i == 0:
            greatest_divisor = i

    return greatest_divisor

print(find_gcd(arr))
```

```python
# A number is an Armstrong number
# if the sum of each digit raised to the number of digits equals the original number.

# n = 153
# d = 3

# 1^3 + 5^3 + 3^3 = 1 + 125 + 27 = 153 == 153 --> Armstrong
```

```python
n = 153
temp = n
d = len(str(n))
sm = 0

while temp > 0:
    digit = temp % 10
    sm += digit ** d
    temp //= 10

print(sm)

if sm == n:
    print("Armstrong Number")
else:
    print("Not Armstrong Number")
```
