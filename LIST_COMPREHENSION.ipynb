```python
numbers = [1,2,3,4,5,6,7,8]
# stores the squares of the even numbers and ignore the odd numbers.
# 1-with and without list coprehension.


# without list coprehension.
sq = []
for num in numbers:
  if num%2==0:
    sq.append(num**2)
sq
```

```python
# wit list coprehension.
even_sq = [num**2 for num in numbers if num%2==0]
even_sq
```

```python
# We can use nested conditional expressions when there are three categories.
# Marks >= 80 --> "Distinction"
# Marks >= 40 --> "pass"
# Otherwise --> "Fail"

marks = [95,72,38,81,45,29]
categorise = ["Distinction" if m >= 80
              else "pass" if m >= 40
              else "Fail"
              for m in marks
              ]
categorise
```

```python
# Problem statements: Given employee salaries,calculate bonuses using these rules:
# salary >= 80,000:20% bonus
# salary >= 50,000 but below 80,000: 10% bonus
# salary below 50,000: 5% bonus
# Return a list containing the bonus amounts, not the updated salaries.

salary = [90000,65000,40000,100000,50000]
bonuses = [s*0.2 if s >= 80000
             else s*0.1 if s >= 50000
             else s*0.05
              for s in salary]
bonuses
```

```python
# Classify online orders by value
# problem statement:Given a list of order values,assign shipping categories using these rules:
# order value >= 2,000 "Free Express shipping"
# order value >= 1,000 but below 2,000 "Free standard shipping"
# order value >= 1,000 "Shipping fee applicable"
orders = [2500,1500,800,3000,1200,500]
```

```python
category = ["Free Express shipping" if v >= 2000
            else "Free standard shipping" if v >= 1000
            else "Shipping fee applicable"
            for v in orders]
category
```

```python
# Given two lists,find elements that are present in the first list but not in the second list.
# List 1:[10,20,30,40]
# List 2:[20,40,50]
# ans = [10,30]

n = [10,20,30,40]
m = [20,40,50]
ans = []
for e in n:
  if e not in m:
    ans.append(e)

print(ans)
```
