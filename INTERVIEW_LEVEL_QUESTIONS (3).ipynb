```python
# Problem statement: an array contains distinct numbers from 1 to n,
# but one number is missing. find the missing number.

# Input: arr = [1,2,4,5,6],n = 6
# output: 3
```

```python
def find_missing(arr,n):
    expected_sum = n*(n+1)//2
    actual_sum = sum(arr)

    return expected_sum-actual_sum

arr = [1,2,4,5,6]
print(find_missing(arr,6))
```

```python
# Find two numbers whose sum equals a tarrget

# problem statements: Given an aaray and a target value,
# return the indices of two distinct elements whose sum equals to target.
# Assume exactly one valid pair exits.

# Input: arr =[2,7,11,15],target = 9
# Output:[0,1]
```

```python
def two_sum(arr, target):
    seen = {}

    for i, num in enumerate(arr):
        required = target - num

        if required in seen:
            return [seen[required], i]

        # store index instead of 1
        seen[num] = i

    return []

arr = [2, 7, 11, 15]
print(two_sum(arr, 9))  # Output: [0, 1]
```

```python
def is_palindrdrome(text):
  cleaned = ""

  for char in text.lower():
    if char.isalnum():
      cleaned+=char

  left = 0
  right = len(cleaned) - 1

  while left < right:
    if cleaned[left] != cleaned[right]:
      return False

    left+=1
    right-=1

  return True

print(is_palindrdrome("A man a plane,a canal:panama"))
print(is_palindrdrome("Hello"))
```

```python
# problem statemets:Two strings are anagrams
# if they contain the same characters with the the same frequencies,regardless of orders.
# Ignore space.

# Input:
# s1:"Listen"
# s2:"Silent"

# output:"Yes"
```

```python
def are_anagrams(s1, s2):
    s1 = s1.replace(" ", "").lower()
    s2 = s2.replace(" ", "").lower()

    if len(s1) != len(s2):
        return False

    frequency = {}

    for char in s1:
        frequency[char] = frequency.get(char, 0) + 1

    for char in s2:
        if char not in frequency:
            return False

        frequency[char] -= 1

        if frequency[char] < 0:
            return False

    return True


print(are_anagrams("Listen", "Silent"))
print(are_anagrams("Hello", "World"))
```

```python
# problem statement: Given s string,find the first character that appears only exactly ones.
# Return -1 if they character repeats.

# Input: "aabbcddee"
# output:'c'
```

```python
def first_unique_character(text):
    frequency = {}

    for char in text:
        frequency[char] = frequency.get(char, 0) + 1

    for char in text:
        if frequency[char] == 1:
            return char

    return -1


print(first_unique_character("aabbcddee"))
print(first_unique_character("aabb"))
```

```python
# Problem statement:Given a sentence,return its longest word.
# if multiply word have the same maximum length, return thr first one.

# Input:"python makes problem solving easier"
# output:"problem"
```

```python
def longest_word(sentence):
    words = sentence.split()

    if not words:
        return ""

    longest = words[0]

    for word in words[1:]:
        if len(word) > len(longest):
            longest = word

    return longest


sentence = "Python makes problem solving easier"
print(longest_word(sentence))
```

```python
# Problem statement:
# Given two aarays sorted in ascending order,
# merge them into one sorted array without sorting the combine result afterward.

# Input:[1,4,7,9] and [2,3,8,10]
# output:[1,2,3,4,7,8,9,10]
```

```python
def merge_sorted_arrays(a, b):
    i = 0
    j = 0
    result = []

    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i])
            i += 1
        else:
            result.append(b[j])
            j += 1

    while i < len(a):
        result.append(a[i])
        i += 1

    while j < len(b):
        result.append(b[j])
        j += 1

    return result


a = [1, 4, 7, 9]
b = [2, 3, 8, 10]

print(merge_sorted_arrays(a, b))
```

```python
# Find the product of all elements except the current element

# Probrlem statement:create a result array in which each position contains
# the product of all the other array elements,without dividing by the current element.

# Example: [1,2,3,4]
# Output:  [24,12,8,6]

# Logic:
# pick an element:
# inner loop: find prd of all, and skip that elem using index.
```

```python
arr = [1,2,3,4]

product = []

for i in range(len(arr)):
  p = 1
  for j in range(len(arr)):
    if i != j:
      p*=arr[j]

  product.append(p)

print(product)
```

```python
# # Answer Multiple Range-Sum Queries

# **Problem statement:** Given an array and several queries, each containing a starting index L and an ending index R,
# calculate the sum of elements from index L to index R inclusive

# **Example:**Array: `[3, 5, 2, 8, 4]`

# - Query L = 1, R = 3 → 15
# - Query L = 0, R = 2 → 10
```

```python
arr = [3, 5, 2, 8, 4]

prefix = [0]
for x in arr:
    prefix.append(prefix[-1] + x)

print(prefix)

def range_sum(L, R):
    return prefix[R + 1] - prefix[L]

print(range_sum(1, 3))
print(range_sum(0, 2))
```

```python
# # Find the Longest Subarray with Equal Numbers of 0s and 1s

# **Problem statement:** Given an array containing only 0 and 1, find the length of the
# longest contiguous subarray containing an equal number of zeros and ones.

# **Example:**

# Input: `[0, 1, 0, 0, 1, 1, 0]`

# Output: `6`
```

```python
array = [0, 1, 0, 0, 1, 1, 0]

longest = 0

for i in range(len(array)):
    zeros = 0
    ones = 0

    for j in range(i, len(array)):
        if array[j] == 0:
            zeros = zeros + 1
        else:
            ones = ones + 1

        if zeros == ones:
            length = j - i + 1
            if length > longest:
                longest = length

print(longest)
```

```python
# Find the Minimum Length Subarray with Sum at Least S

# Problem statement: Given an array of positive integers and a target S,
# find the minimum length of a contiguous subarray whose sum is at least S.
# Print 0 if no such subarray exists.

# Example: [2, 3, 1, 2, 4, 3], S = 7
# output: [4,3] --> 2
```

```python
array = [3,2,1,2,4,3]
s = 7

min_len = 7

for i in range(len(array)):
  total = 0
  for j in range(i,len(array)):
    total = total +array[j]

    if total >= s:
      length = j-i+1
      if length < min_len:
        min_len = length
      break

print(min_len)
```
